````python
from google.cloud import bigquery

# 1. 실제 GCP 콘솔에 표시된 소문자 Project ID 입력
PROJECT_ID = "sns-analysis-prj"

client = bigquery.Client(project=PROJECT_ID)

# 2. `프로젝트ID.데이터셋.테이블` 전체 경로 지정
sql = f"""
    SELECT * 
    FROM `{PROJECT_ID}.sns_analysis.test`
"""

# 판다스 데이터프레임으로 변환
df = client.query(sql).to_dataframe()

print(df.head())
````

#### 쿼리로 데이터 업데이트 하는 방법

```sql
UPDATE 테이블명
SET 컬럼1 = 변경할_값1, 컬럼2 = 변경할_값2
WHERE 조건절;
```

````python
# UPDATE 실행
sql = f"""
    UPDATE `{PROJECT_ID}.sns_analysis.test`
    SET gender = '여자'
    WHERE gender = 'f'
"""

# DML 쿼리는 결과 완료 대기(result)를 호출해야 작업 완료가 보장됩니다.
query_job = client.query(sql)
query_job.result() 

print(f"업데이트 완료! 변경된 행 수: {query_job.num_dml_affected_rows}")
````
