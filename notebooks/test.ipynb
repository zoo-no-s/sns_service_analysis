```python
from google.cloud import bigquery

# 1. 실제 GCP 콘솔에 표시된 소문자 Project ID 입력
PROJECT_ID = "sns-analysis-prj"  # 실제 ID로 변경 (예: sns-analysis-prj-123456)

client = bigquery.Client(project=PROJECT_ID)

# 2. `프로젝트ID.데이터셋.테이블` 전체 경로 지정
sql = f"""
    SELECT * 
    FROM `{PROJECT_ID}.sns_analysis.test`
"""

# 판다스 데이터프레임으로 변환
df = client.query(sql).to_dataframe()

print(df.head())
```
